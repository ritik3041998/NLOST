```python
import os
import shutil

# Input and output folders
input_folder = r"C:\Users\Ritik Madhasia\OneDrive\Desktop\nlos\3D\obj"
output_folder = r"C:\Users\Ritik Madhasia\OneDrive\Desktop\nlos\3D\obj_out"

# Create output folder if not exists
os.makedirs(output_folder, exist_ok=True)

for filename in os.listdir(input_folder):
    if filename.endswith(".mat"):
        # Remove "_recon" from filename
        new_name = filename.replace("_recon", "")
        
        src_path = os.path.join(input_folder, filename)
        dst_path = os.path.join(output_folder, new_name)
        
        # Copy and rename file
        shutil.copy(src_path, dst_path)

        print(f"Renamed: {filename} -> {new_name}")

print("Done.")
```
